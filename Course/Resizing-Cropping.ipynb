```python
import cv2
import cvzone
import os

script_dir = os.path.dirname(os.path.abspath(os.getcwd()))
img_path = os.path.join(script_dir, "Images", "Pets.jpg")
img = cv2.imread(img_path)

try:
    if img is not None:

        resized = cv2.resize(img, (300, 280))
        cropped = img[280:300,200:320]

        cv2.imshow(mat=img, winname=f"Pets | Shape = {img.shape} | Height = {img.shape[0]} | Width = {img.shape[1]}")
        cv2.imshow(mat=resized, winname=f"Reshaped Pets | Shape = {resized.shape} | Height = {resized.shape[0]} | Width = {resized.shape[1]}")
        cv2.imshow(mat=cropped, winname=f"Cropped Pets | Shape = {cropped.shape} | Height = {cropped.shape[0]} | Width = {cropped.shape[1]}")

        cv2.waitKey(0)
        cv2.destroyAllWindows()
    else:
        print(f"Failed to Display Images, They don't Exist")
except BaseException as e:
    print(f"Failed to Display Images, An Error has Occured = {e}")            
```
