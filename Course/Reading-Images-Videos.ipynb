```python
import cv2
import cvzone
import os

script_dir = os.path.dirname(os.path.abspath(os.getcwd()))
image1 = os.path.join(script_dir, "Images", "Apricots.jpg")
image2 = os.path.join(script_dir, "Images", "Pets.jpg")
image3 = os.path.join(script_dir, "Images", "Toys.jpg")

video1 = os.path.join(script_dir, "Videos", "Bus.mp4")
video2 = os.path.join(script_dir, "Videos", "Chef.mp4")
video3 = os.path.join(script_dir, "Videos", "Construction.mp4")

apricots = cv2.imread(image1)
pets = cv2.imread(image2)
toys = cv2.imread(image3)

try:
    if apricots is not None and pets is not None and toys is not None:
        cv2.imshow(mat=apricots, winname="Apricots")
        cv2.imshow(mat=pets, winname="Pets")
        cv2.imshow(mat=toys, winname="Toys")
        
        cv2.waitKey(0)
        cv2.destroyAllWindows()
    else:
        print("Failed to Display Images, They don't Exist")   
except BaseException as e:
    print(f"Failed to Display Images = {e}")

try:    
    bus = cv2.VideoCapture(video1)
    chef = cv2.VideoCapture(video2)
    construction = cv2.VideoCapture(video3)

    while True:
        success1, img1 = bus.read()
        success2, img2 = chef.read()
        success3, img3 = construction.read()

        cv2.imshow("Bus", img1)
        cv2.imshow("Chef", img2)
        cv2.imshow("Construction", img3)

        if cv2.waitKey(1) & 0xFF == ord('q'):
            break 
except BaseException as e:
    print(f"Failed to Access & View your Videos, An Error has Occured = {e}")        
```
