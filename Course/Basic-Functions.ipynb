```python
import cv2
import cvzone
import os
import numpy as np

script_dir = os.path.dirname(os.path.abspath(os.getcwd()))
img_path = os.path.join(script_dir, "Images", "Apricots.jpg")

img = cv2.imread(img_path) # Apricots Image
kernel = np.ones((5,5),np.uint8) # Kernel Matrix

try:
    if img is not None:

        img_gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY) # Turns the Image Gray
        blurred_image = cv2.GaussianBlur(img, ksize=(45,45), sigmaX=0) # Blurs the Image
        img_edges = cv2.Canny(img, threshold1=120, threshold2=240) # Highlights the Edges of Objects in the Image
        dialated_image = cv2.dilate(img_edges, kernel=kernel, iterations=1) # Thickens the Edges
        eroded_image = cv2.erode(dialated_image, kernel=kernel, iterations=1) # Smoothens the Edges
        
        cv2.imshow(mat=img, winname="Normal Apricots")
        cv2.imshow(mat=img_gray, winname="Grayed Apricots")
        cv2.imshow(mat=blurred_image, winname="Blurred Apricots")
        cv2.imshow(mat=img_edges, winname="Edged Apricots")
        cv2.imshow(mat=dialated_image, winname="Dialated Apricots")
        cv2.imshow(mat=eroded_image, winname="Eroded Apricots")
        
        cv2.waitKey(0)
        cv2.destroyAllWindows()

    else:
        print("Failed to Load Images, They doesn't Exist")
except BaseException as e:
    print(f"Failed to Display Images, An Error has Occured = {e}")        
```
