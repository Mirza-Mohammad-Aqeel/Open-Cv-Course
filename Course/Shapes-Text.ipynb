```python
import cv2
import cvzone
import os
import numpy as np
script_dir = os.path.dirname(os.path.abspath(os.getcwd()))

img_path = os.path.join(script_dir, "Images", "Apricots.jpg")
img = cv2.imread(img_path)

try:
    if img is not None:

        cv2.imshow(winname="Apricots", mat=img)
        cv2.line(img,(10,19), (img.shape[1], img.shape[0]), (0,213,198), (5))
        cv2.rectangle(img, (13,14), (45,67), (100,230,45), (5), cv2.FILLED)
        cv2.circle(img, (67), (12), (21,21,255), (5), cv2.FILLED)
        cv2.putText(img, "These are Apricots", (89,91), cv2.FONT_HERSHEY_PLAIN, fontScale=(2.31), thickness=(6))

        cv2.waitKey(0)
        cv2.destroyAllWindows()
    else:
        print(f"Failed to Display Image, It doesn't Exist")
except BaseException as e:
    print(f"Failed to Display Image, An Error has Occured = {e}")           
```
