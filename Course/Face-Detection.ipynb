```python
import cv2
import os
script_dir = os.path.dirname(os.path.abspath(os.getcwd()))

face_cascade_path = cv2.data.haarcascades + "haarcascade_frontalface_default.xml"
faceCascade = cv2.CascadeClassifier(face_cascade_path)
img_path = os.path.join(script_dir, "Images","Faces.jpg")
faces_image = cv2.imread(img_path)
imgGray = cv2.cvtColor(faces_image, cv2.COLOR_BGR2GRAY)

faces = faceCascade.detectMultiScale(imgGray, 1.1, 6)
for (x,y,w,h) in faces:
    cv2.rectangle(faces_image, (x,y), (x+w,y+h), color=(34, 124, 227), thickness=5)

cv2.imshow(winname="Faces", mat=faces_image)
cv2.waitKey(0)
```
