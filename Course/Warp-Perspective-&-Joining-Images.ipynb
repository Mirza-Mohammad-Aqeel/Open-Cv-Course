```python
import cv2
import cvzone
import os
import numpy as np
script_dir = os.path.dirname(os.path.abspath(os.getcwd()))

img_path1 = os.path.join(script_dir, "Images", "Apricots.jpg")
img_path2 = os.path.join(script_dir, "Images", "Toys.jpg")

img1 = cv2.imread(img_path1)
img2 = cv2.imread(img_path2)
img3 = cv2.cvtColor(img2, cv2.COLOR_BGR2GRAY)
points1 = np.float32([[34,56],[89,94], [13,25], [27, 36]])
points2 = np.float32([[42,69],[31,38], [123,135], [11,21]])
matrix = cv2.getPerspectiveTransform(points1, points2)
imgOutput = cv2.warpPerspective(img1, matrix, (210,310))

try:
    if img1 is not None:
        cv2.imshow(mat=imgOutput, winname="Apricots")
        cv2.waitKey(0)
        cv2.destroyAllWindows()
    else:
        print(f"Failed to Display Image, It doesn't Exist")
except BaseException as e:
    print(f"Failed to Display Image, An Error has Occured = {e}") 

hor = np.hstack((img1, img2, img1))
ver = np.vstack((img2, img3, img2))
try:
    if img1 is not None and img2 is not None:
        cv2.imshow(winname="Horizontal",mat=hor)
        cv2.imshow(winname="Vertical",mat=ver)
        cv2.waitKey(0)
        cv2.destroyAllWindows()        
    else:
        print(f"Failed to Display Images,  They don't Exist")
except BaseException as e:
    print(f"Failed to Display Graphs, An Error has Occured = {e}")                    
```
